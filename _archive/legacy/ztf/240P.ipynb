```python
# Fetch vector data (Cartesian coordinates)
vec = obj.vectors()[0]
vec_frag = obj_frag.vectors()[0]

# Extract x, y, z coordinates
p1 = [vec['x'], vec['y'], vec['z']]
p2 = [vec_frag['x'], vec_frag['y'], vec_frag['z']]
```

```python
import numpy as np

# Convert to numpy arrays for vector subtraction
pos1 = np.array([vec['x'], vec['y'], vec['z']])
pos2 = np.array([vec_frag['x'], vec_frag['y'], vec_frag['z']])

# Calculate distance in AU
dist_au = np.linalg.norm(pos1 - pos2)

# Convert to kilometers (1 AU ≈ 149,597,870.7 km)
dist_km = dist_au * 149597870.7

print(f"Physical Distance: {dist_km:.2f} km")
```
